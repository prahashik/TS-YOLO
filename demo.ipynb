```python
import gradio as gr
import torch
from ultralytics import YOLO

# Load TS-YOLO pipeline / model
model = YOLO('yolov8n.pt')

def predict(image):
    results = model(image)
    res_plotted = results[0].plot()
    return res_plotted

demo = gr.Interface(fn=predict, inputs=gr.Image(type='pil'), outputs=gr.Image(type='pil'), title='TS-YOLO Pipeline Demo')
demo.launch(share=True)
```
